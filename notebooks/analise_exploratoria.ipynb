```python
import pandas as pd
```

```python
path_orders = '../data/raw/olist_orders_dataset.csv'
path_items = '../data/raw/olist_order_items_dataset.csv'
path_products = '../data/raw/olist_products_dataset.csv'
```

```python
df_orders = pd.read_csv(path_orders)
df_items = pd.read_csv(path_items)
df_products = pd.read_csv(path_products)
```

```python
print("--- Tabela PEDIDOS (Orders) ---")
display(df_orders.head())
print("\nInformações Técnicas (Tipos de Dados):")
df_orders.info()

print("\n" + "="*50 + "\n")

print("--- Tabela ITENS (Items) ---")
display(df_items.head())
print("\nInformações Técnicas:")
df_items.info()

print("\n" + "="*50 + "\n")

print("--- Tabela PRODUTOS (Products) ---")
display(df_products.head())
print("\nInformações Técnicas:")
df_products.info()
```

```python
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
print(df_orders.dtypes)
```

```python
import matplotlib.pyplot as plt
```

```python
df_orders['month_year'] = df_orders['order_purchase_timestamp'].dt.to_period('M')

orders_per_month = df_orders.groupby('month_year').size()

plt.figure(figsize=(12, 6))
orders_per_month.plot(kind='line', marker='o', color='b')
plt.title('Evolução do Número de Pedidos (2016-2018)')
plt.xlabel('Mês/Ano')
plt.ylabel('Quantidade de Pedidos')
plt.grid(True)
plt.show()
```

```python
df_merge_1 = pd.merge(df_orders, df_items, on='order_id', how='inner')

df_completo = pd.merge(df_merge_1, df_products, on='product_id', how='inner')

print(f"Tamanho da Tabela Final: {df_completo.shape}")
display(df_completo.head())
```

```python
import seaborn as sns
```

```python
receita_por_categoria = df_completo.groupby('product_category_name')['price'].sum().sort_values(ascending=False)
top_10_categorias = receita_por_categoria.head(10)

plt.figure(figsize=(10, 6))
sns.barplot(x=top_10_categorias.values, y=top_10_categorias.index, palette='viridis')
plt.title('Top 10 Categorias por Receita Total')
plt.xlabel('Receita (R$)')
plt.ylabel('Categoria')
plt.show()

print("Top 3 Categorias Campeãs:")
print(top_10_categorias.head(3))
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(x=df_completo['price'])

plt.title('Distribuição de Preços dos Produtos (Boxplot)')
plt.xlabel('Preço (R$)')
plt.show()


print("Estatísticas de Preço:")
print(df_completo['price'].describe())
```

```python
df_limpo_peso = df_completo.dropna(subset=['product_weight_g'])

plt.figure(figsize=(10, 6))

sns.scatterplot(x='product_weight_g', y='freight_value', data=df_limpo_peso, alpha=0.3)

plt.title('Correlação: Peso do Produto x Valor do Frete')
plt.xlabel('Peso (gramas)')
plt.ylabel('Frete (R$)')
plt.show()

correlacao = df_limpo_peso['product_weight_g'].corr(df_limpo_peso['freight_value'])
print(f"Coeficiente de Correlação: {correlacao:.2f}")
```

```python
df_completo['order_delivered_customer_date'] = pd.to_datetime(df_completo['order_delivered_customer_date'])


df_completo['delivery_days'] = (df_completo['order_delivered_customer_date'] - df_completo['order_purchase_timestamp']).dt.days

plt.figure(figsize=(10, 6))
sns.histplot(df_completo['delivery_days'].dropna(), bins=50, kde=True, color='green')

plt.title('Distribuição do Tempo de Entrega (Dias)')
plt.xlabel('Dias para Entregar')
plt.ylabel('Frequência')
plt.xlim(0, 50) 
plt.show()

print(f"Tempo médio de entrega: {df_completo['delivery_days'].mean():.1f} dias")
```

```python
df_completo['order_purchase_timestamp'] = pd.to_datetime(df_completo['order_purchase_timestamp'])
df_completo['order_delivered_customer_date'] = pd.to_datetime(df_completo['order_delivered_customer_date'])

df_completo['delivery_days'] = (df_completo['order_delivered_customer_date'] - df_completo['order_purchase_timestamp']).dt.days


plt.figure(figsize=(10, 6))


sns.histplot(df_completo['delivery_days'].dropna(), bins=50, kde=True, color='green')

plt.title('Distribuição do Tempo de Entrega (em Dias)')
plt.xlabel('Dias para Entregar')
plt.ylabel('Quantidade de Pedidos')
plt.xlim(0, 60) 
plt.show()


media_entrega = df_completo['delivery_days'].mean()
print(f"A média de entrega do e-commerce é de {media_entrega:.1f} dias.")
```

```python

colunas_finais = [
    'order_id', 'order_status', 'order_purchase_timestamp', 'order_delivered_customer_date',
    'delivery_days', 'product_category_name', 'price', 'freight_value', 
    'product_weight_g', 'customer_id' 
]


df_export = df_completo[colunas_finais].copy()


df_export.to_csv('olist_dados_analise_final.csv', index=False)

print("Arquivo 'olist_dados_analise_final.csv' criado com sucesso! ")
```
